```python
#importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
#import cleaned dataset
df = pd.read_csv('cleaned-home-data.csv')
```

```python
 df['SalePrice(in Crores)'].describe()
```

```python
plt.figure(figsize=(8,5))

plt.boxplot(df['SalePrice(in Crores)'])

plt.ylabel('Sale Price (Crores)')
plt.title('Sale Price Box Plot')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Carpet Area'], df['SalePrice(in Crores)'])

plt.xlabel('Carpet Area')
plt.ylabel('Sale Price (Crores)')
plt.title('Carpet Area vs Sale Price')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.boxplot(x='BHK', y='SalePrice(in Crores)', data=df)

plt.xlabel('BHK')
plt.ylabel('Sale Price (Crores)')
plt.title('BHK vs Sale Price')

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(x='Location', y='SalePrice(in Crores)', data=df)

plt.xlabel('Location')
plt.ylabel('Sale Price (Crores)')
plt.title('Location vs Sale Price')
plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Bathroom', y='SalePrice(in Crores)', data=df)

plt.xlabel('Bathroom')
plt.ylabel('Sale Price (Crores)')
plt.title('Bathroom vs Sale Price')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x='FlatFloor', y='SalePrice(in Crores)', data=df)

plt.xlabel('Flat Floor')
plt.ylabel('Sale Price (Crores)')
plt.title('Flat Floor vs Sale Price')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Transaction', y='SalePrice(in Crores)', data=df)

plt.xlabel('Transaction')
plt.ylabel('Sale Price (Crores)')
plt.title('Transaction vs Sale Price')

plt.show()
```
